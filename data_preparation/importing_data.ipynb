```python
from datasets import load_dataset
```

```python
train_ds = load_dataset("nampdn-ai/tiny-textbooks", split="train").shuffle(seed=123)
test_ds = load_dataset("nampdn-ai/tiny-textbooks", split="test").shuffle(seed=123)
```

```python
print(train_ds)
print(test_ds)
```

```python
train_ds = train_ds["text"]
test_ds = test_ds["text"]
```

```python
print(len(train_ds))
print(len(test_ds))
```

```python
total_samples = 200000

train_samples = 170000
val_samples = 20000
test_samples = 10000

val_ds = train_ds[train_samples:train_samples + val_samples]
train_ds = train_ds[:train_samples]
test_ds = test_ds[:test_samples]
```

```python
total = len(train_ds) + len(val_ds) + len(test_ds)

print(len(train_ds))
print(len(val_ds))
print(len(test_ds))
print(f"Train: {len(train_ds) / total}% | Val: {len(val_ds) / total}% | Test: {len(test_ds) / total}")
```

```python
with open("data/train_data", "w", encoding="utf-8") as f:
    for text in train_ds:
        f.write(f"{text} <|endoftext|>")
```

```python
with open("data/val_data", "w", encoding="utf-8") as f:
    for text in val_ds:
        f.write(f"{text} <|endoftext|>")
```

```python
with open("data/test_data", "w", encoding="utf-8") as f:
    for text in test_ds:
        f.write(f"{text} <|endoftext|>")
```
